# Interactive Schwarzschild Orbits
This notebook implements a numerical lab for exploring orbits around a Schwarzschild black hole, replicating the "Manipulate" functionality from the RelativityToolkit.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
import ipywidgets as widgets
from IPython.display import display
```

## 1. Physics: The Geodesic Equations
In the Schwarzschild metric, the orbits of a particle are governed by the effective potential:
$V_{eff}(r) = \sqrt{(1 - 2M/r)(1 + L^2/r^2)}$

However, to simulate the actual path, we solve the equations of motion for $u = 1/r$ with respect to $\phi$:
$\frac{d^2u}{d\phi^2} + u = \frac{M}{L^2} + 3Mu^2$

```python
def geodesic_eq(phi, y, M, L):
    """
    y[0] = u = 1/r
    y[1] = du/dphi
    """
    u, v = y
    dudphi = v
    dvdphi = (M / L**2) - u + 3 * M * u**2
    return [dudphi, dvdphi]

def solve_orbit(M, L, r0, v_r0):
    """
    r0: initial radius
    v_r0: initial radial velocity (dr/dphi)
    """
    # Safety: L cannot be 0 for this specific u(phi) formulation
    L = max(L, 0.1)
    
    u0 = 1.0 / r0
    v0 = - (1.0 / r0**2) * v_r0
    
    # Event 1: Stop if we hit the event horizon (u = 1/2M)
    def event_horizon(phi, y, M, L):
        return y[0] - 1.0/(2.0*M)
    event_horizon.terminal = True
    
    # Event 2: Stop if we escape to infinity (u = 0)
    def escape(phi, y, M, L):
        return y[0]
    escape.terminal = True
    
    phi_span = (0, 100 * np.pi) 
    phi_eval = np.linspace(0, 100 * np.pi, 2000)
    
    # Using 'LSODA' for stiffness handling near the horizon
    sol = solve_ivp(geodesic_eq, phi_span, [u0, v0], args=(M, L), 
                    t_eval=phi_eval, events=[event_horizon, escape], 
                    method='LSODA', rtol=1e-5)
    
    r = 1.0 / np.maximum(sol.y[0], 1e-10) # Prevent division by zero
    phi = sol.t
    
    # Convert to Cartesian for plotting
    x = r * np.cos(phi)
    y = r * np.sin(phi)
    
    return x, y
```

## 2. Interactive Plotting

```python
def update_plot(M, L, r0, vr0):
    # Clear any existing figures to prevent memory leaks and slow-downs
    plt.close('all')
    
    fig = plt.figure(figsize=(8, 8))
    
    # Solve
    try:
        x, y = solve_orbit(M, L, r0, vr0)
        plt.plot(x, y, label="Orbit Path", color='cyan', lw=1.5)
        
        # Draw the Event Horizon
        horizon_theta = np.linspace(0, 2*np.pi, 100)
        plt.plot(2*M*np.cos(horizon_theta), 2*M*np.sin(horizon_theta), 'r--', label="Event Horizon (2M)", alpha=0.8)
        plt.fill(2*M*np.cos(horizon_theta), 2*M*np.sin(horizon_theta), 'black', alpha=0.4)
        
    except Exception as e:
        plt.text(0, 0, f"Orbit Error: {str(e)}", ha='center', color='white', backgroundcolor='red')

    plt.axhline(0, color='grey', lw=0.5, alpha=0.5)
    plt.axvline(0, color='grey', lw=0.5, alpha=0.5)
    plt.title(f"Schwarzschild Orbit: M={M}, L={L}")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.legend(loc='upper right')
    plt.axis('equal')
    
    # Set limits based on r0 to keep the view stable
    limit = r0 * 2.0
    plt.xlim(-limit, limit)
    plt.ylim(-limit, limit)
    plt.grid(True, which='both', linestyle='--', alpha=0.3)
    plt.show()

# Create widgets with manual update trigger potential
m_slider = widgets.FloatSlider(value=1.0, min=0.1, max=5.0, step=0.1, description='Mass (M)', continuous_update=True)
l_slider = widgets.FloatSlider(value=4.3, min=0.1, max=10.0, step=0.1, description='Ang. Mom (L)', continuous_update=True)
r0_slider = widgets.FloatSlider(value=15.0, min=5.0, max=50.0, step=1.0, description='Start r', continuous_update=True)
vr0_slider = widgets.FloatSlider(value=0.0, min=-1.0, max=1.0, step=0.05, description='Start vr', continuous_update=True)

# Optional: Add a button to force update if the "automatic" release isn't working
update_btn = widgets.Button(description="Update Plot", button_style='info')

def on_button_clicked(b):
    update_plot(m_slider.value, l_slider.value, r0_slider.value, vr0_slider.value)

update_btn.on_click(on_button_clicked)

# Display UI
ui = widgets.VBox([m_slider, l_slider, r0_slider, vr0_slider, update_btn])
out = widgets.interactive_output(update_plot, {'M': m_slider, 'L': l_slider, 'r0': r0_slider, 'vr0': vr0_slider})

display(ui, out)
```
